## 4. Análise e Discussão

a) Sobre Ponto Branco e Constância de Cor (Seção 4.1.11): Por que a correção de ponto branco é considerada uma "transformação de adaptação cromática"?

    R: Por que essa correção ajusta a percepção de cor do cérebro humano, mantendo o branco constante sob diferentes iluminações.


b) Espaços Independentes vs. Dependentes (Seção 4.1.12 e 4.2.3): Qual a principal diferença prática entre o espaço de cores CIE XYZ e o sRGB? Por que não exibimos imagens diretamente em XYZ nos monitores comuns?

    R: O CIE XYZ é um espaço criado para ser preciso, cobrindo todo o espectro de luz visível pelo olho humano. O sRGB é um espaço menor, feito para telas que usam píxeis que combinam as cores azul, verde e vermelho. Sem isso, cada monitor interpretaria os sinais XYZ de um jeito, resultando em cores inconsistentes.


c) A Necessidade da Correção Gama (Seção 4.1.13): Durante a aula, aplicamos a correção gama Ccorrected=C(1/γ) linear . O que aconteceria com a percepção das sombras em uma imagem se essa correção fosse omitida?

    R: As sombras da imagem ficariam muito mais escuras e perderiam muitos detalhes. Isso acontece porque os monitores não exibem o brilho de forma linear, então essa correção é sempre necessária.


d) Percepção Humana e o Modelo CIELAB (Seção 4.1.14): O modelo L∗a∗b∗ é frequentemente chamado de "perceptualmente uniforme". O que isso significa em termos de distância entre cores?

    R: é chamado de "perceptualmente uniforme" porque foi projetado para que a distância entre dois pontos no espaço de cores corresponda à diferença de cor que o olho humano realmente percebe.


e) Engenharia de Vídeo e YCbCr (Seção 4.3.4): No experimento de visualização, os canais Cb e Cr parecem muito mais "suaves" ou "borrados" que o canal Y . Como a compressão de vídeo (como JPEG ou MPEG) utiliza essa característica para reduzir o tamanho dos arquivos?

    R: A compressão de vídeo utiliza essa característica com base no fato de que o olho humano é muito mais sensível a variações de brilho do que a variações de cor. Então, podemos aplicar transformações onde alguns detalhes são descartadas, especialmente nos canais borrados. Também é aplicada a quantização com mais agressividade nos canais de croma do que o de luminância, economizando muitos bits. 


f) Representação Cilíndrica e Robustez (Seção 4.2.3): Ao realizar a segmentação de uma região com cor uniforme, por que o canal H (Hue) do modelo HSV é frequentemente preferido em relação aos canais do modelo RGB quando a imagem possui variações de iluminação (sombras e reflexos)?

    R: O canal H do modelo HSV é preferido para segmentação de cor nesse contexto porque ele separa a informação de cor da luminância, tornando a detecção mais robusta.
